# Load the data

Place the files in the ../../05_src/data/fires/ directory and load the appropriate file. 

```python
# Load environment variables.

%load_ext dotenv
%dotenv

# Load the libraries as required.

import os
import pandas as pd
import numpy as np
import pickle

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
```

```python
# Load data
columns = [
    'coord_x', 'coord_y', 'month', 'day', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain', 'area' 
]
fires_dt = (pd.read_csv('../../05_src/data/fires/forestfires.csv', header = 0, names = columns))
fires_dt.info()
```

There are no missing values in the data set (i.e., 517 non-null). Data types are as expected; no conversions required. `month` and `day` are listed as object (e.g., strings like "apr", "fri"). `coord_x` and `coord_y` are integers (int64). All other variables (e.g., `ffmc`, `temp`, `wind`, `rain`, `area`, etc.) are decimals (float64).

```python
# View the first few rows of the data.

print(fires_dt.head())

# View the summary statistics.

fires_dt.describe()
```

Based on the first few rows of data, it appears that the `rain` is on a much smaller scale compared to `ffmc` and `dc`. The summary statistics show that the minimum value for `ffmc` is 18.7, the average is 90.6, and the maximum is 96.2 and for `dc` the minimum value is 7.9, the average is 547.9, and the maximum is 860.6. In comparison, for `rain` the minimum, 25%, 50%, and 75% are all 0.0 and the maximum is 6.4. 

Hence, we will need to scale the variables during preprocessing. By ensuring that the variables are all on the the same scale, we will ensure that the model does not artificially consider `ffmc` or  `dc` to be more important than `rain` just becasue it's scale is bigger.

Next, the mean of most variables is similar to the median (i.e., 50%). For example, for `ffmc` the mean is 90.6 and the median is 91.6 or for `temp` the mean is 18.89 and the median is 19.30. For these variables, the data appears to be normally distributed. 

However, for `area` the median is 0.52 ha compared to the mean at 12.85 ha. This appears to be a right-skewed distribution where bigger areas being burned is creating a tail. This is also the case for `rain` where what seems to be the outlier of 6.4 mm/m2 is pulling the rest of the points at 0.0 mm/m2 towards the right. Lastly, there also appears to be a left-skewed distribution for `dc` where the mean is 547.94 but the median is much higher at 664.20.

Considering the skewed distributions of some of these variables, we likely would need to log-transform the variables to bring in the outlier values closer to the rest of the data points.

# Get X and Y

Create the features data frame and target data.

```python
# Create features dataframe.

X = fires_dt.drop(columns = ["area"])

# Create target data.

Y = fires_dt["area"]
```

```python
# Split into testing and training sets.

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
```

# Preprocessing

Create two [Column Transformers](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html), called preproc1 and preproc2, with the following guidelines:

- Numerical variables

    * (Preproc 1 and 2) Scaling: use a scaling method of your choice (Standard, Robust, Min-Max). 
    * Preproc 2 only: 
        
        + Choose a transformation for any of your input variables (or several of them). Evaluate if this transformation is convenient.
        + The choice of scaler is up to you.

- Categorical variables: 
    
    * (Preproc 1 and 2) Apply [one-hot encoding](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html) where appropriate.


+ The only difference between preproc1 and preproc2 is the non-linear transformation of the numerical variables.
    


```python
# Define the numeric variables.

num_cols = X_train.select_dtypes(include=['int64', 'float64']).columns
print(f"The numerical columns are: {num_cols.tolist()}")

# Define the categorical variables.

cat_cols = X_train.select_dtypes(include=['object']).columns
print(f"The categorical columns are: {cat_cols.tolist()}")
```

### Preproc 1

Create preproc1 below.

+ Numeric: scaled variables, no other transforms.
+ Categorical: one-hot encoding.

```python
# Create preproc1.

preproc1 = ColumnTransformer(
    transformers=[
        ('num_transform', StandardScaler(), num_cols),
        ('cat_transform', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ], remainder='drop'
)
```

### Preproc 2

Create preproc1 below.

+ Numeric: scaled variables, non-linear transformation to one or more variables.
+ Categorical: one-hot encoding.

```python
# Since the distributions of the variables are skewed, I will apply yeo-johnson transformation to the numeric variables.

pipe_num_yj = Pipeline([
    ('imputer', SimpleImputer(strategy = 'median')),
    ('standardizer', StandardScaler()),
    ('transform', PowerTransformer(method='yeo-johnson'))
])

# Create preproc2.

preproc2 = ColumnTransformer([
    ('num_transform', pipe_num_yj, num_cols),
    ('cat_transform', OneHotEncoder(handle_unknown='ignore'), cat_cols)
], remainder='drop')
```

## Model Pipeline


Create a [model pipeline](https://scikit-learn.org/stable/modules/generated/sklearn.pipeline.Pipeline.html): 

+ Add a step labelled `preprocessing` and assign the Column Transformer from the previous section.
+ Add a step labelled `regressor` and assign a regression model to it. 

## Regressor

+ Use a regression model to perform a prediction. 

    - Choose a baseline regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Choose a more advance regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Both model choices are up to you, feel free to experiment.

```python
# Pipeline A = preproc1 + baseline

Pipeline_A  = Pipeline([
    ("preprocessing", preproc1),
    ("regressor", Ridge(random_state=42))
])
```

```python
# Pipeline B = preproc2 + baseline

Pipeline_B  = Pipeline([
    ("preprocessing", preproc2),
    ("regressor", Ridge(random_state=42))
])
```

```python
# Pipeline C = preproc1 + advanced model

Pipeline_C  = Pipeline([
    ("preprocessing", preproc1),
    ("regressor", RandomForestRegressor(random_state=42))
])
```

```python
# Pipeline D = preproc2 + advanced model

Pipeline_D  = Pipeline([
    ("preprocessing", preproc2),
    ("regressor", RandomForestRegressor(random_state=42))
])
```

# Tune Hyperparams

+ Perform GridSearch on each of the four pipelines. 
+ Tune at least one hyperparameter per pipeline.
+ Experiment with at least four value combinations per pipeline.

```python
# Define the parameter grid for Ridge regression.

rdg_param_grid = {
    'regressor__alpha': [0.001, 0.003, 0.01, 0.03, 0.1, 0.3, 1, 3, 10, 30, 100, 300, 1000]
}

# Define the parameter grid for Random Forest Regressor.

rf_param_grid = {
    'regressor__n_estimators': [10, 30, 50, 100, 200, 300]
}

# Define the error metric.

scoring_metric = ['neg_root_mean_squared_error']
```

```python
# Run GridSearchCV on Pipeline A.

Grid_A = GridSearchCV(
    estimator=Pipeline_A, 
    param_grid=rdg_param_grid, 
    cv=5,
    scoring=scoring_metric
)
Grid_A.fit(X_train, y_train)

print(f"Grid A Best Score (Negative RMSE): {Grid_A.best_score_:.4f}")
print(f"Grid A Best Params: {Grid_A.best_params_}")
```

```python
# Run GridSearchCV on Pipeline B.

Grid_B = GridSearchCV(
    estimator=Pipeline_B, 
    param_grid=rdg_param_grid, 
    cv=5,
    scoring=scoring_metric
)
Grid_B.fit(X_train, y_train)

print(f"Grid B Best Score (Negative RMSE): {Grid_B.best_score_:.4f}")
print(f"Grid B Best Params: {Grid_B.best_params_}")
```

```python
# Run GridSearchCV on Pipeline C.

Grid_C = GridSearchCV(
    estimator=Pipeline_C, 
    param_grid=rf_param_grid, 
    cv=5,
    scoring=scoring_metric
)
Grid_C.fit(X_train, y_train)

print(f"Grid C Best Score (Negative RMSE): {Grid_C.best_score_:.4f}")
print(f"Grid C Best Params: {Grid_C.best_params_}")
```

```python
# Run GridSearchCV on Pipeline D.

Grid_D = GridSearchCV(
    estimator=Pipeline_D, 
    param_grid=rf_param_grid, 
    cv=5,
    scoring=scoring_metric
)
Grid_D.fit(X_train, y_train)

print(f"Grid D Best Score (Negative RMSE): {Grid_D.best_score_:.4f}")
print(f"Grid D Best Params: {Grid_D.best_params_}")
```
